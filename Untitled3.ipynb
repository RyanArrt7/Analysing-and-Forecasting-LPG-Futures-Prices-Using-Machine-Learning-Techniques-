```python
import nbformat
import sys

# List of notebooks to merge, in order
notebook_files = ['Untitled.ipynb', 'Untitled1.ipynb','Untitled2.ipynb', 'ensemble.ipynb']

# Create a new, empty notebook to hold the merged cells
merged_notebook = nbformat.v4.new_notebook()

try:
    # Loop through each notebook file and read its contents
    for file in notebook_files:
        print(f"Reading notebook: {file}...")
        with open(file, 'r', encoding='utf-8') as f:
            nb = nbformat.read(f, as_version=4)
            # Append all cells from the current notebook to the merged notebook
            merged_notebook.cells.extend(nb.cells)

    # Write the combined notebook to a new file
    output_filename = 'merged_notebook.ipynb'
    with open(output_filename, 'w', encoding='utf-8') as f:
        nbformat.write(merged_notebook, f)

    print(f"Success! The merged notebook has been saved to '{output_filename}'")

except FileNotFoundError as e:
    print(f"Error: A notebook file was not found. Please check the file path: {e.filename}", file=sys.stderr)
except Exception as e:
    print(f"An unexpected error occurred: {e}", file=sys.stderr)
```
